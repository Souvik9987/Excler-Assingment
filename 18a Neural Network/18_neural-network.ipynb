```python
# Case Study: SONAR — Detecting Mines vs. Rocks
# 1️⃣ Business Objective
# Goal:
# To build an intelligent system that can automatically detect whether an underwater sonar signal is reflected from a metallic mine (potentially dangerous) or a harmless rock.
# This is vital for:
# • Maritime safety: Prevent ships and submarines from colliding with mines.
# • Naval defense: Identify and safely remove underwater mines.
# • Resource exploration: Distinguish between useful metal structures and natural seabed objects.
```

```python
# Ouitlier Capping 
for i in df.select_dtypes(include = ['int','float']).columns:
    q1 = df[i].quantile(0.25)
    q3 = df[i].quantile(0.75)

    IQR = q1 - q3 
    lower_bound = q1 - 1.5*IQR
    upper_bound = q3 + 1.5*IQR

    df[i] = df[i].clip(lower_bound, upper_bound)
```

```python
# 2. Model Implementation
# ● Construct a basic ANN model using your chosen high-level neural network library. Ensure your model includes at least one hidden layer.
# ● Divide the dataset into training and test sets.
# ● Train your model on the training set and then use it to make predictions on the test set.
```

```python
# Construction of ANN Model 

from keras.models import Sequential 
from keras.layers import Dense 

from tensorflow.keras.callbacks import EarlyStopping 
early = EarlyStopping(monitor = 'val_accuracy', patience = 4)

# Creation of model having multi layer 
model = Sequential()
model.add(Dense(12, input_dim =60, activation = 'relu')) # 1st Layer Input layer

model.add(Dense(8, activation = 'relu')) # 2nd Layer or 1st Hidden Layer 
model.add(Dense(8, activation = 'relu')) # 2nd Hiddenn layer
model.add(Dense(4, activation = 'relu')) # 3rd Hidden layer

model.add(Dense(1, activation = 'sigmoid'))  # Output layer
```

```python
# Target columns encoding 
import pandas as pd

# Create encoded target DataFrame
y_df = pd.DataFrame()

y_df["Y_encoded"] = df["Y"].map({
    "R": 0,
    "M": 1
})

# Add encoded column to original DataFrame
df = pd.concat([df, y_df], axis=1)

# Remove original categorical target
df.drop("Y", axis=1, inplace=True)
```

```python
# Separate features and target
x = df.drop("Y_encoded", axis=1)
y = df["Y_encoded"]
```

```python
# Divide the dataset into training and testin 
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state= 42)
```

```python
# Model compiling 
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# ● Train your model on the training set and then use it to make predictions on the test set.

# Fit the Model 
history = model.fit(x,y, validation_split = 0.33, epochs = 50, batch_size = 10)
```

```python
# Make predictions on test data
y_pred_prob = model.predict(x_test)

print(y_pred_prob[:10])

# Convert Problities into classification
y_pred = (y_pred_prob >= 0.5).astype(int)
```

```python
# Compare with actual valus

compare = pd.DataFrame({
    "Actual Value" : y_test,
    "Predicted Value" : y_pred.flatten()
})
compare.head(10)
```

```python
# ============================================================
# ANN HYPERPARAMETER TUNING USING GRID SEARCH
# Binary Classification
# ============================================================
import warnings
warnings.filterwarnings("ignore")
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import GridSearchCV


# ------------------------------------------------------------
# 1. Create ANN model
# ------------------------------------------------------------

def create_model(optimizer="adam", neurons=8):

    model = Sequential()

    # Input + Hidden Layer
    model.add(
        Dense(
            neurons,
            input_dim=x_train.shape[1],
            activation="relu"
        )
    )

    # Second Hidden Layer
    model.add(Dense(8, activation="relu"))

    # Output Layer - Binary Classification
    model.add(Dense(1, activation="sigmoid"))

    # Compile
    model.compile(
        optimizer=optimizer,
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ------------------------------------------------------------
# 2. Wrap ANN for GridSearchCV
# ------------------------------------------------------------

model = KerasClassifier(
    model=create_model,
    verbose=0
)


# ------------------------------------------------------------
# 3. Define parameters for Grid Search
# ------------------------------------------------------------

param_grid = {
    "model__neurons": [8, 12, 16],
    "model__optimizer": ["adam", "rmsprop"],
    "batch_size": [10, 20],
    "epochs": [30, 50]
}


# ------------------------------------------------------------
# 4. Apply Grid Search
# ------------------------------------------------------------

grid = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=3,
    scoring="accuracy",
    n_jobs=1
)


# ------------------------------------------------------------
# 5. Run Grid Search on training data
# ------------------------------------------------------------

grid_result = grid.fit(
    x_train,
    y_train
)


# ------------------------------------------------------------
# 6. Display best parameters
# ------------------------------------------------------------

print("Best Parameters:")
print(grid_result.best_params_)

print("\nBest Cross-Validation Accuracy:")
print(grid_result.best_score_)


# ------------------------------------------------------------
# 7. Get the best ANN model
# ------------------------------------------------------------

best_model = grid_result.best_estimator_


# ------------------------------------------------------------
# 8. Evaluate best model on TEST data
# ------------------------------------------------------------

test_accuracy = best_model.score(
    x_test,
    y_test
)

print("\nTest Accuracy:")
print(test_accuracy)


# ------------------------------------------------------------
# 9. Make predictions on test data
# ------------------------------------------------------------

y_pred = best_model.predict(x_test)

print("\nFirst 10 Predictions:")
print(y_pred[:10])
```
