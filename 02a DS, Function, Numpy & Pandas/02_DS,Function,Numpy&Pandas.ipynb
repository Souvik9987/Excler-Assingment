```python
# 2. Create a tuple that stores 3 student names. Try changing the second name in the tuple.
#     What happens? Explain why.

# A tuple is immutable, which means its elements cannot be changed, added or removed after the tuple is 
# created.
```

```python
# Create a tuple
students = ("Jerry", "Tom", "Spike")

print("Original Tuple:", students)

# Try changing the second name
students = list(students)
students[1] = "Tanjiro"
students = tuple(students)

print("Updated Tuple:", students)
```

```python
# 6.	Given a 'marks' variable, print the grade:
# - 90 and above: 'A'
# - 75–89: 'B'
# - 50–74: 'C'
# - Below 50: 'Fail'
```

```python
marks = int(input("Enter your marks "))
if marks >= 90:
    print("A")
elif marks >= 75 and  marks <= 89 : 
    print("B")
elif marks >= 50 and marks <= 74:
    print("C")
else:
    print("Fail")
```

```python
# 7.	Given a number, check if it's positive, and if it is also even. 
# If not positive, print if it's zero or negative.
```

```python
# Given number
num = 5

# Check the number
if num > 0:
    print("The number is positive.")

    if num % 2 == 0:
        print("The number is also even.")
    else:
        print("The number is also odd.")

elif num == 0:
    print("The number is zero.")

else:
    print("The number is negative.")
```

```python
# 8. Create:
# - A scalar using np.array(5)
# - A 1D array with values 1 to 5
# - A 2D array (2x3) with values from 10 to 60 in steps of 10
```

```python
import numpy as np

# 1. Scalar (0D array)
scalar = np.array(5)

# 2. 1D array with values 1 to 5
array_1d = np.array([1, 2, 3, 4, 5])

# 3. 2D array (2x3) with values from 10 to 60
array_2d = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

# Print the arrays
print("Scalar:")
print(scalar)

print("\n1D Array:")
print(array_1d)

print("\n2D Array:")
print(array_2d)
```

```python
# 11. Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. 
# Then:
# - Use .info(), .describe()
# - Select 'Name' and 'City' columns
# - Drop 'City' column
# - Fill any missing values in 'Salary' column with the mean
# - Remove any duplicate rows
```
