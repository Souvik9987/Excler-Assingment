```python
# 2.Feature Extraction:
    
# Decide on the features that will be used for computing similarity (e.g., genres, user ratings).
# Convert categorical features into numerical representations if necessary.
# Normalize numerical features if required.
```

```python
# Feature Extraction 
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer(stop_words = 'english')

genre_matrix = vectorizer.fit_transform(df['genre'])
```

```python
# 3.Recommendation System:

# Design a function to recommend anime based on cosine similarity.
# Given a target anime, recommend a list of similar anime based on cosine similarity scores.
# Experiment with different threshold values for similarity scores to adjust the recommendation list size.
# Analyze the performance of the recommendation system and identify areas of improvement.
```

```python
# Compute cosine Similarity 
from sklearn.metrics.pairwise import cosine_similarity
cosin_sim = cosine_similarity(genre_matrix)
```

```python
# Create index mapping 
indices = pd.Series(df.index, index = df['name']).drop_duplicates()
```

```python
# Recommendation Function 
def recommend_anime (anime_name, top_n = 10):

    if anime_name not in indices:
        return "Anime not found!!"

        
    idx = indices[anime_name]
    similarity_scores = list(enumerate(cosin_sim[idx]))

    similarity_scores = sorted(
        similarity_scores,
        key = lambda x: x[1],
        reverse = True
    )

    similarity_scores = similarity_scores[0:top_n+1]
    anime_indices = [i[0] for i in similarity_scores]

    return df.iloc[anime_indices][['name','genre','rating']].sort_values('rating',ascending = False)
```

```python
# Recommended Anime 
recommend_anime('Kimi no Na wa.')
```

```python
# Treshold Recommendation 

def recommend_treshold(anime_name, treshold = 0.4):
    if anime_name not in indices:
        return "Anime Not Found!!!"

    idx = indices[anime_name]
    similarity_scores = list(enumerate(cosin_sim[idx]))

    filtered = []

    for i , score in similarity_scores:

        if score >= treshold and i != idx:
            filtered.append((i, score))

    filtered = sorted(filtered,
                     key = lambda x: x[1],
                     reverse = True
                     )
    anime_indices = [i[0] for i in filtered]


    return df.iloc[anime_indices][
        ['name','genre','rating','episodes']
    ]
```

```python
recommend_treshold("Kimi no Na wa.", 0.05)
```
